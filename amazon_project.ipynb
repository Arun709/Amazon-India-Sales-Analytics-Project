```python
import pandas as pd
path = r"C:\Users\kanna\Downloads\amazon_india_complete_2015_2025.csv.csv"
raw = pd.read_csv(path, dtype=str, usecols=['order_date'])
s = raw['order_date'].astype(str)

# 1) DD-MM-YYYY and DD/MM/YYYY
mask_dd = s.str.match(r'^\d{2}[-/]\d{2}[-/]\d{4}$', na=False) & pd.to_datetime(s, dayfirst=True, errors='coerce').notna()
parsed_dd = pd.to_datetime(s[mask_dd], format=None, dayfirst=True, errors='coerce')

# 2) MM/DD/YYYY (month-first)
mask_md = s.str.match(r'^\d{2}/\d{2}/\d{4}$', na=False) & pd.to_datetime(s, dayfirst=False, errors='coerce').notna()
parsed_md = pd.to_datetime(s[mask_md], format='%m/%d/%Y', errors='coerce')

# 3) Explicitly assign back into a clean column
clean = pd.Series(pd.NaT, index=s.index)
clean.loc[mask_dd] = parsed_dd
clean.loc[mask_md] = parsed_md

# 4) For any remaining, try common explicit formats
remaining = clean.isna()
formats = ['%d-%m-%Y','%d/%m/%Y','%m/%d/%Y','%d%m%Y','%Y-%m-%d']
for fmt in formats:
    if remaining.any():
        parsed = pd.to_datetime(s[remaining], format=fmt, errors='coerce')
        clean.loc[remaining] = parsed
        remaining = clean.isna()

# 5) Result check
clean.isna().sum(), clean.dropna().head()
```

```python
import pandas as pd

# Function to clean and standardize dates
def clean_date(date_str):
    try:
        return pd.to_datetime(date_str, errors='coerce').strftime('%Y-%m-%d')
    except:
        return None

# Apply cleaning to order_date
df['order_date'] = df['order_date'].apply(clean_date)
df = df.dropna(subset=['order_date'])  # Drop rows with invalid dates
```

```python
def clean_price(price):
    if isinstance(price, str):
        # Remove ₹, commas, and other non-numeric characters
        price = price.replace('₹', '').replace(',', '').strip()
        try:
            return float(price)
        except:
            return None
    return float(price) if price else None

df['original_price_inr'] = df['original_price_inr'].apply(clean_price)
df = df.dropna(subset=['original_price_inr'])  # Drop invalid prices
```

```python
def clean_rating(rating):
    if pd.isna(rating):
        return None
    if isinstance(rating, str):
        if '/' in rating:
            num, denom = rating.split('/')
            return float(num) / float(denom) * 5.0  # Convert to 1-5 scale
        return float(rating.strip(' stars'))
    return float(rating)

df['customer_rating'] = df['customer_rating'].apply(clean_rating)
# Impute missing ratings (e.g., with median)
df['customer_rating'] = df['customer_rating'].fillna(df['customer_rating'].median())
```

```python
import pandas as pd
import numpy as np

# Updated function to clean delivery_days
def clean_delivery_days(value):
    if pd.isna(value):
        return None
    if isinstance(value, str):
        value = value.lower().strip()
        if value == 'same day':
            return 0
        if '-' in value:
            # Handle ranges like '1-2 days'
            try:
                parts = value.split('-')
                low, high = map(float, [p.replace('days', '').strip() for p in parts[0:2]])
                return (low + high) / 2
            except:
                return None
        if 'days' in value:
            # Handle formats like '2 days'
            try:
                return float(value.replace('days', '').strip())
            except:
                return None
        try:
            return float(value)
        except:
            return None
    if isinstance(value, (int, float)):
        if value < 0 or value > 30:  # Handle negative and unrealistic values
            return None
        return value
    return None

# Apply cleaning
df['delivery_days'] = df['delivery_days'].apply(clean_delivery_days)

# Impute missing values with median of valid delivery days
median_days = df['delivery_days'].median()
df['delivery_days'] = df['delivery_days'].fillna(median_days)

# Convert to integer
df['delivery_days'] = df['delivery_days'].astype(int)

# Verify cleaning
print(df['delivery_days'].describe())
print(df['delivery_days'].head())
```

```python
# Calculate IQR for original_price_inr
Q1 = df['original_price_inr'].quantile(0.25)
Q3 = df['original_price_inr'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['original_price_inr'] < lower_bound) | (df['original_price_inr'] > upper_bound)]
print(f"Number of price outliers: {len(outliers)}")

# Category-specific price caps
price_caps = {
    'Smartphones': 150000,
    'Laptops': 300000
}

# Correct outliers
def correct_price(row):
    price = row['original_price_inr']
    category = row['subcategory']
    if price > upper_bound:
        corrected_price = price / 100  # Fix decimal point errors
        cap = price_caps.get(category, 150000)
        if corrected_price <= cap:
            return corrected_price
        return cap
    return price

df['original_price_inr'] = df.apply(correct_price, axis=1)

# Validate with final_amount_inr
df['original_price_inr'] = df['original_price_inr'].where(
    df['original_price_inr'] >= df['final_amount_inr'],
    df['final_amount_inr'] / (1 - df['discount_percent'] / 100)
)

# Verify
print(df['original_price_inr'].describe())
```
